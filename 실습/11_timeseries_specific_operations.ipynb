````python
import yfinance as yf
import pandas as pd
from matplotlib import pyplot as plt
````

````python
# 구글 주식 데이터 로딩
goog = yf.download("GOOG", start="2004-01-01", end="2024-12-31")
goog.head()
````

````python
# [+] 종가(Close) 데이터만 사용하기
goog = goog['Close']
goog
````

````python
# [+] Lineplot 시각화
goog.plot()
````

#### <strong> 리샘플링(Resampling)
+ ```resample()```: 빈도를 기준으로 집계 연산
+ ```asfreq()```: 빈도 단위 별 데이터 선택

````python
# [+] 다운샘플링: resample()
# 일별 주가를 월별 평균 주가로 집계합니다.
goog_downsampled = goog.resample('ME').mean() # [cite: 1367]
goog_downsampled
````

````python
fig, ax = plt.subplots(2, 1, sharey=True, figsize=(12, 6))
goog.plot(label='original', ax=ax[0])
ax[0].legend()
goog_downsampled.plot(label='downsampled', ax=ax[1])
ax[1].legend()
plt.tight_layout()
````

````python
# [+] 다운샘플링: asfreq()
# 월별 말일의 주가 데이터 포인트만 선택합니다.
goog_downsampled = goog.asfreq('ME') # [cite: 1396]
print(goog_downsampled)

goog_downsampled.plot()
````

````python
# [+] 업샘플링: resample()
# 일별 주가를 12시간 단위 평균으로 빈도를 늘립니다 (NaN 발생).
goog_upsampled = goog.resample('12h').mean() # [cite: 1428]
print(goog_upsampled)
````

#### <strong> 시프팅(Shifting)
+ ```shift(periods)``` : ```periods``` 값 만큼 시계열 이동

````python
# [+] 900일 뒤로 시프팅 적용
goog_shifted = goog.shift(periods=900) # [cite: 1460]
goog_shifted
````

````python
fig, ax = plt.subplots(2, 1, sharey=True, figsize=(12, 6))

# Legends and annotations
local_max = pd.to_datetime('2007-11-05')
offset = pd.Timedelta(900, 'D')

goog.plot(label='original', ax=ax[0])
ax[0].legend()
ax[0].get_xticklabels()[2].set(weight='heavy', color='red')
ax[0].axvline(local_max, alpha=0.3, color='red')

goog_shifted.plot(label='shifted 900 days', ax=ax[1])
ax[1].legend()
ax[1].get_xticklabels()[2].set(weight='heavy', color='red')
ax[1].axvline(local_max + offset, alpha=0.3, color='red')

plt.tight_layout()
````

````python
# shifting 응용: 주가 변화율 계산
goog_shifted = goog.shift(periods=1) # [+] 시프팅 [cite: 1525]
(goog - goog_shifted) / goog_shifted # [+] 주가 변화율 계산 [cite: 1525]
````

````python
# shifting 응용: 차분 시계열 구하기
goog_diff = goog - goog_shifted  # [+] 차분 (현재 값 - 이전 값) [cite: 1555, 1556]
print(goog_diff)

goog_diff.plot()
````

#### <strong> 윈도잉(Windowing)

````python
# [+] 365일 단위 윈도잉
# 365일을 기준으로 이동 평균(Moving Average) 등의 집계 연산을 수행할 윈도우 객체를 만듭니다.
goog_rolled = goog.rolling(365) # [cite: 1588]
print(goog_rolled)

goog_rolled.mean().plot()
````

````python
data = pd.DataFrame({'input': goog.values.flatten(),
                     # moving average
                     'one-year rolling_mean': goog_rolled.mean().values.flatten(), 
                     # standard deviation
                     'one-year rolling_std': goog_rolled.std().values.flatten()}, index=goog.index)
ax = data.plot(style=['-', '--', ':'])
ax.lines[0].set_alpha(0.3)
````
